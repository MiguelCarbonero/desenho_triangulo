```python
import numpy as np
import matplotlib.pyplot as plt

trianguloOrigi = np.array([
    [0,2,4,0],
    [0,4,0,0],
    [1,1,1,1]
])

def pltTransf(original, transformado, titulo):
  plt.figure(figsize = (6,6))
  plt.axhline(0, color = 'blue', linewidth = 1)
  plt.axvline(0, color = 'blue', linewidth = 1)
  plt.grid(True, linestyle='--', alpha = 0.5)

  plt.plot(original[0,:], original[1,:], 'b--', label = 'Original')
  plt.plot(transformado[0,:], transformado[1,:], 'r--', label = 'Transformado')
  plt.xlim(-5,10)
  plt.ylim(-5,10)

  plt.title(titulo)
  plt.legend()
  plt.show()
```

```python
tx, ty = 3, 2

matrizTrans = np.array([
    [1,0,tx],
    [0,1,ty],
    [0,0,1]
])

trianguloTrans = matrizTrans @ trianguloOrigi

pltTransf(trianguloOrigi, trianguloTrans, "Translação (tx = 3, ty = 2)")
```

```python
#Escala

sx,sy = 2.0,1.5

matrizEsc = np.array([
    [sx,0,0],
    [0,sy,0],
    [0,0,1]
])

trianguloEsc = matrizEsc @ trianguloOrigi

pltTransf(trianguloOrigi, trianguloEsc, "Escala (sx = 2.0, sy = 1.5)")
```

```python
theta_graus = 190
theta = np.radians(theta_graus)

matrizRot = np.array([
    [np.cos(theta), -np.sin(theta), 0],
    [np.sin(theta), np.cos(theta), 0],
    [0,0,1]
])

trianguloRot = matrizRot @ trianguloOrigi

pltTransf(trianguloOrigi, trianguloRot, f"Rotação ({theta_graus}º)")
```
